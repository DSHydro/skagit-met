```python
import geopandas as gpd
mask = gpd.read_file('../data/GIS/SkagitBoundary.json')
mask
```

```python
mask.explore()
```

```python
gf = gpd.read_file('../data/GIS/SkagitSubBasin_HUC8.geojson')
gf = gf[gf.Name.isin(['Upper Skagit', 'Lower Skagit', 'Sauk'])]
gf
```

```python
m = gf.explore(column='Name', cmap='tab10')
mask.explore(m=m,
          style_kwds={
        'fillOpacity': 0,  # Set fill opacity to 0 to remove fill
        'color': 'blue',   # Set outline color (e.g., blue)
        'weight': 3        # Set outline weight (e.g., 3 pixels)
    }
)
```
