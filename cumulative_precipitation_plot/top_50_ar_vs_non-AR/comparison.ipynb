```python
# Top 50 prism_3d_tot with ar_scale = 0 (non-AR events)
non_ar_top50 = events[events['ar_scale'] == 0].nlargest(50, 'prism_3d_tot')['date'].tolist()
print("Top 50 non-AR events (ar_scale = 0):")
print(non_ar_top50)
```

```python
# Top 50 prism_3d_tot with ar_scale > 0 (AR events)
ar_top50 = events[events['ar_scale'] > 0].nlargest(50, 'prism_3d_tot')['date'].tolist()
print("Top 50 AR events (ar_scale > 0):")
print(ar_top50)
```
